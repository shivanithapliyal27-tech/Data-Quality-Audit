```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
df = pd.read_excel(
    '/content/sample_-_superstore.xls',
    sheet_name='Orders'
)

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

display(df.head())
```

```python

# Check column names
print("Column Names:")
print(df.columns.tolist())

# Check data types and non-null counts
print("\nDataset Information:")
df.info()

# Statistical summary
display(df.describe(include='all').T)
```

```python

missing_audit = pd.DataFrame({
    'Missing_Count': df.isnull().sum(),
    'Missing_Percentage': (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

missing_audit = missing_audit.sort_values(
    by='Missing_Count',
    ascending=False
)

display(missing_audit)
```

```python

duplicate_count = df.duplicated().sum()

print("Total Duplicate Rows:", duplicate_count)

# Display duplicate records for review
duplicate_rows = df[df.duplicated(keep=False)]

display(duplicate_rows.head(10))
```

```python

# Check numeric columns
numeric_columns = ['Sales', 'Quantity', 'Discount', 'Profit']

for col in numeric_columns:
    if col in df.columns:
        print(f"\n--- {col} ---")
        print("Minimum:", df[col].min())
        print("Maximum:", df[col].max())
        print("Missing:", df[col].isnull().sum())

if 'Sales' in df.columns:
    print("\nNegative Sales:", (df['Sales'] < 0).sum())

if 'Quantity' in df.columns:
    print("Zero or Negative Quantity:",
          (df['Quantity'] <= 0).sum())

if 'Discount' in df.columns:
    print("Discount Outside 0-1:",
          ((df['Discount'] < 0) |
           (df['Discount'] > 1)).sum())
```

```python

issues = []

# 1. Missing values
missing_total = int(df.isnull().sum().sum())

issues.append({
    'Issue': 'Missing Values',
    'Issue_Count': missing_total,
    'Severity': 'High' if missing_total > 0 else 'None',
    'Recommended_Action': 'Investigate and handle missing values'
})

# 2. Exact duplicate rows
duplicate_total = int(df.duplicated().sum())

issues.append({
    'Issue': 'Duplicate Rows',
    'Issue_Count': duplicate_total,
    'Severity': 'Medium' if duplicate_total > 0 else 'None',
    'Recommended_Action': 'Review exact duplicates before removal'
})

# 3. Negative sales
negative_sales = (
    int((df['Sales'] < 0).sum())
    if 'Sales' in df.columns else 0
)

issues.append({
    'Issue': 'Negative Sales',
    'Issue_Count': negative_sales,
    'Severity': 'Review' if negative_sales > 0 else 'None',
    'Recommended_Action': 'Investigate unusual sales values'
})

# 4. Non-positive quantity
invalid_quantity = (
    int((df['Quantity'] <= 0).sum())
    if 'Quantity' in df.columns else 0
)

issues.append({
    'Issue': 'Non-positive Quantity',
    'Issue_Count': invalid_quantity,
    'Severity': 'Review' if invalid_quantity > 0 else 'None',
    'Recommended_Action': 'Validate quantity values'
})

# Create issue log
issue_log = pd.DataFrame(issues)

display(issue_log)
```

```python

df_cleaned = df.drop_duplicates().copy()

print("Original Rows:", len(df))
print("Cleaned Rows:", len(df_cleaned))
print("Rows Removed:", len(df) - len(df_cleaned))

display(df_cleaned.head())
```

```python

# Save cleaned dataset
df_cleaned.to_csv(
    'Cleaned_Sample.csv',
    index=False
)

# Save issue log
issue_log.to_csv(
    'Data_Quality_Issue_Log.csv',
    index=False
)

# Save missing values audit
missing_audit.to_csv(
    'Missing_Values_Audit.csv'
)

# Save range audit
range_audit = []

for col in ['Sales', 'Quantity', 'Discount', 'Profit']:
    if col in df.columns:
        range_audit.append({
            'Column': col,
            'Minimum': df[col].min(),
            'Maximum': df[col].max(),
            'Missing_Count': int(df[col].isnull().sum())
        })

pd.DataFrame(range_audit).to_csv(
    'Data_Range_Audit.csv',
    index=False
)

print("All audit files saved successfully!")
```

```python

checklist = pd.DataFrame([
    {
        'Check': 'Missing Values',
        'Status': 'PASS' if df.isnull().sum().sum() == 0 else 'REVIEW'
    },
    {
        'Check': 'Exact Duplicate Rows',
        'Status': 'PASS' if df.duplicated().sum() == 0 else 'REVIEW'
    },
    {
        'Check': 'Negative Sales',
        'Status': 'PASS' if (
            'Sales' not in df.columns or (df['Sales'] < 0).sum() == 0
        ) else 'REVIEW'
    },
    {
        'Check': 'Non-positive Quantity',
        'Status': 'PASS' if (
            'Quantity' not in df.columns or (df['Quantity'] <= 0).sum() == 0
        ) else 'REVIEW'
    },
    {
        'Check': 'Discount Range',
        'Status': 'PASS' if (
            'Discount' not in df.columns or
            ((df['Discount'] >= 0) & (df['Discount'] <= 1) |
             df['Discount'].isnull()).all()
        ) else 'REVIEW'
    }
])

display(checklist)

checklist.to_csv('Data_Quality_Checklist.csv', index=False)
```
